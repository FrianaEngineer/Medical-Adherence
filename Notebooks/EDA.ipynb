## DESYNPUF_ID uniqueness

Check whether beneficiary IDs are unique within each file and whether Sample 1 and Sample 2 share the same IDs.

```python
def check_id_uniqueness(csv_path):
    """Return row count, unique ID count, and whether DESYNPUF_ID is unique per row."""
    ids = pd.read_csv(csv_path, usecols=['DESYNPUF_ID'], dtype=str)['DESYNPUF_ID']
    n_rows = len(ids)
    n_unique = ids.nunique()
    return {
        'file': csv_path.name,
        'rows': n_rows,
        'unique_ids': n_unique,
        'duplicates': n_rows - n_unique,
        'is_unique': n_rows == n_unique,
    }
```

```python
def collect_ids(folder):
    """Collect all unique DESYNPUF_ID values from every CSV in a sample folder."""
    all_ids = set()
    for csv_path in sorted(folder.glob('*.csv')):
        ids = pd.read_csv(csv_path, usecols=['DESYNPUF_ID'], dtype=str)['DESYNPUF_ID']
        all_ids.update(ids)
    return all_ids

sample_1_ids = collect_ids(SAMPLE_1_DIR)
sample_2_ids = collect_ids(SAMPLE_2_DIR)

overlap = sample_1_ids & sample_2_ids

print(f'Unique IDs in Sample 1: {len(sample_1_ids):,}')
print(f'Unique IDs in Sample 2: {len(sample_2_ids):,}')
print(f'Overlapping IDs: {len(overlap):,}')
print(f'Same ID sets: {sample_1_ids == sample_2_ids}')
```

### Are `PROD_SRVC_ID` values repeated per patient?

Each PDE row is one fill. For a given patient, we compare **total fills** vs **unique `PROD_SRVC_ID`** values. If fills > unique products, the same drug code appears on more than one row (a refill).

```python
PATIENT_ID = "DESYNPUF_ID"
PDE_PATH = SAMPLE_1_DIR / "DE1_0_2008_to_2010_Prescription_Drug_Events_Sample_1.csv"

pde = pd.read_csv(
    PDE_PATH,
    usecols=[PATIENT_ID, "PDE_ID", "PROD_SRVC_ID", "SRVC_DT"],
    dtype={PATIENT_ID: str, "PDE_ID": str, "PROD_SRVC_ID": str},
    low_memory=False,
)

patient_products = pde.groupby(PATIENT_ID, observed=True).agg(
    n_fills=(PATIENT_ID, "count"),
    n_unique_products=("PROD_SRVC_ID", "nunique"),
)
patient_products["n_repeat_fills"] = patient_products["n_fills"] - patient_products["n_unique_products"]
patient_products["has_repeat_product"] = patient_products["n_fills"] > patient_products["n_unique_products"]

n_patients = len(patient_products)
n_with_repeats = patient_products["has_repeat_product"].sum()

print(f"PDE fills: {len(pde):,}")
print(f"Patients with PDE: {n_patients:,}")
print(f"Unique PROD_SRVC_ID overall: {pde['PROD_SRVC_ID'].nunique():,}")
print()
print(f"Patients with at least one repeated PROD_SRVC_ID: {n_with_repeats:,} ({100 * n_with_repeats / n_patients:.1f}%)")
print(f"Patients where every fill is a different product: {(~patient_products['has_repeat_product']).sum():,} ({100 * (1 - n_with_repeats / n_patients):.1f}%)")
print()
print("Per-patient summary (fills vs unique products):")
display(patient_products.describe().round(1))

# Among repeated patient-product pairs: different fill dates vs same day
repeat_pairs = (
    pde.groupby([PATIENT_ID, "PROD_SRVC_ID"], observed=True)
    .agg(n_fills=("SRVC_DT", "count"), n_unique_dates=("SRVC_DT", "nunique"))
    .reset_index()
)
repeat_pairs = repeat_pairs[repeat_pairs["n_fills"] > 1]
print(f"\nRepeated patient-product pairs: {len(repeat_pairs):,}")
print(f"  Different fill dates: {(repeat_pairs['n_unique_dates'] > 1).sum():,} ({100 * (repeat_pairs['n_unique_dates'] > 1).mean():.1f}%)")
print(f"  Same fill date only: {(repeat_pairs['n_unique_dates'] == 1).sum():,}")

# Example: patient with the most repeat fills
if n_with_repeats:
    example_id = patient_products.sort_values("n_repeat_fills", ascending=False).index[0]
    example = pde[pde[PATIENT_ID] == example_id].sort_values("SRVC_DT")
    print(f"\nExample patient {example_id}: {len(example)} fills, {example['PROD_SRVC_ID'].nunique()} unique products")
    display(example["PROD_SRVC_ID"].value_counts().head(5).to_frame("fill_count"))
```

### Map `PROD_SRVC_ID` to FDA NDC and check drug-level repeats

[`PROD_SRVC_ID`](https://open.fda.gov/apis/drug/ndc/how-to-use-the-endpoint/) is an 11-digit NDC-style code. We:

1. Convert it to **product NDC** (`#####-####`, first 9 digits) and **package NDC** (`#####-####-##`)
2. Query the [openFDA Drug NDC API](https://open.fda.gov/apis/drug/ndc/download/) in batches (cached locally)
3. Compare refill repeats at **package**, **product NDC**, and **FDA generic name** levels

**Note:** DE-SynPUF uses synthetic 2008-era codes; many may not appear in today's FDA directory. Product-level grouping still works without a match.

```python
import json
import ssl
import time
import urllib.error
import urllib.parse
import urllib.request

FDA_NDC_API = "https://api.fda.gov/drug/ndc.json"
FDA_CACHE_PATH = SAMPLE_1_DIR / "fda_ndc_lookup_cache.json"
FDA_BATCH_SIZE = 10           # product_ndc terms per API call
FDA_TOP_PRODUCTS_TO_QUERY = 200  # most common product NDC keys in Sample 1 PDE

_ssl_ctx = ssl.create_default_context()
_ssl_ctx.check_hostname = False
_ssl_ctx.verify_mode = ssl.CERT_NONE


def ndc11_to_product(ndc11: str) -> str:
    s = str(ndc11).zfill(11)
    return f"{s[:5]}-{s[5:9]}"


def ndc11_to_package(ndc11: str) -> str:
    s = str(ndc11).zfill(11)
    return f"{s[:5]}-{s[5:9]}-{s[9:11]}"


def load_fda_cache() -> dict:
    if FDA_CACHE_PATH.exists():
        return json.loads(FDA_CACHE_PATH.read_text())
    return {}


def save_fda_cache(cache: dict) -> None:
    FDA_CACHE_PATH.write_text(json.dumps(cache, indent=2))


def query_fda_product_ndcs(product_ndcs: list[str]) -> list[dict]:
    """Batch OR-search on product_ndc (openFDA requires quoted terms)."""
    if not product_ndcs:
        return []
    search = " OR ".join(f'product_ndc:"{ndc}"' for ndc in product_ndcs)
    params = urllib.parse.urlencode({
        "search": search,
        "limit": min(100, max(10, len(product_ndcs) * 5)),
    })
    url = f"{FDA_NDC_API}?{params}"
    req = urllib.request.Request(url, headers={"User-Agent": "medical-adherence-eda"})
    with urllib.request.urlopen(req, timeout=60, context=_ssl_ctx) as resp:
        payload = json.load(resp)
    if "error" in payload:
        return []
    return payload.get("results", [])


def lookup_fda_products(product_ndcs: list[str], cache: dict) -> dict:
    """Return {product_ndc: {generic_name, brand_name, package_ndcs}} using cache + API."""
    pending = [p for p in product_ndcs if p not in cache]
    for i in range(0, len(pending), FDA_BATCH_SIZE):
        batch = pending[i : i + FDA_BATCH_SIZE]
        try:
            results = query_fda_product_ndcs(batch)
        except (urllib.error.URLError, TimeoutError, json.JSONDecodeError) as exc:
            print(f"FDA API batch failed ({batch[0]} ...): {exc}")
            continue
        for rec in results:
            prod = rec.get("product_ndc")
            if not prod:
                continue
            packages = [p.get("package_ndc") for p in rec.get("packaging", []) if p.get("package_ndc")]
            cache[prod] = {
                "generic_name": rec.get("generic_name"),
                "brand_name": rec.get("brand_name"),
                "package_ndcs": packages,
            }
        for prod in batch:
            cache.setdefault(prod, None)
        time.sleep(0.3)
    return {prod: cache.get(prod) for prod in product_ndcs}


# Quick API sanity check (known real NDC in today's FDA directory)
_api_test = query_fda_product_ndcs(["65862-008"])
print("openFDA API check (65862-008 / metformin):", "OK" if _api_test else "no result")

# --- Attach NDC keys to PDE rows ---
pde_ndc = pde[[PATIENT_ID, "PROD_SRVC_ID"]].copy()
pde_ndc["product_ndc"] = pde_ndc["PROD_SRVC_ID"].map(ndc11_to_product)
pde_ndc["package_ndc"] = pde_ndc["PROD_SRVC_ID"].map(ndc11_to_package)

product_freq = pde_ndc["product_ndc"].value_counts()
top_products = product_freq.head(FDA_TOP_PRODUCTS_TO_QUERY).index.tolist()

fda_cache = load_fda_cache()
fda_lookup = lookup_fda_products(top_products, fda_cache)
save_fda_cache(fda_cache)

matched_products = [p for p, v in fda_lookup.items() if v]
print(f"\nQueried top {len(top_products):,} product NDC keys from Sample 1 PDE")
print(f"FDA matches: {len(matched_products):,} ({100 * len(matched_products) / len(top_products):.1f}% of queried)")

if matched_products:
    display(pd.DataFrame([
        {"product_ndc": p, **fda_lookup[p]} for p in matched_products[:10]
    ]))
else:
    print("No FDA matches for Sample 1 product codes (expected for DE-SynPUF synthetic NDCs).")

pde_ndc["generic_name"] = pde_ndc["product_ndc"].map(
    lambda p: (fda_lookup.get(p) or {}).get("generic_name") if fda_lookup.get(p) else pd.NA
)


def repeat_patient_stats(frame: pd.DataFrame, drug_col: str) -> pd.Series:
    fills = frame.groupby(PATIENT_ID, observed=True).size()
    uniq = frame.groupby(PATIENT_ID, observed=True)[drug_col].nunique()
    return pd.Series({
        "patients": fills.size,
        "mean_fills": round(fills.mean(), 1),
        "mean_unique_drugs": round(uniq.mean(), 1),
        "patients_with_repeat": int((fills > uniq).sum()),
        "pct_with_repeat": round(100 * (fills > uniq).mean(), 1),
    })


rows = []
for label, col in [
    ("Package NDC (PROD_SRVC_ID)", "PROD_SRVC_ID"),
    ("Product NDC (9-digit key)", "product_ndc"),
    ("FDA generic name (matched rows only)", "generic_name"),
]:
    sub = pde_ndc if col != "generic_name" else pde_ndc[pde_ndc["generic_name"].notna()]
    if sub.empty:
        rows.append({"level": label, "patients": 0, "pct_with_repeat": pd.NA, "note": "no FDA matches"})
    else:
        rows.append({"level": label, **repeat_patient_stats(sub, col)})

display(pd.DataFrame(rows))

print("Package = exact same PROD_SRVC_ID refilled")
print("Product = same 9-digit NDC product (different package sizes count as repeats)")
print("Generic = same FDA generic name (subset with API matches)")
print(f"Cache saved to: {FDA_CACHE_PATH}")
```

## Why can PDE unique IDs exceed Carrier 1A + 1B?

### Why this is expected (not a data error)

Your intuition — *"you need a visit to get a prescription"* — applies to **new** prescriptions, but Medicare data does not require a **Part B carrier claim** for every **Part D fill**:

1. **Different claim types** — Carrier = Part B (physician/office). PDE = Part D (pharmacy). They are billed and stored separately.
2. **Refills & renewals** — Ongoing medications can be refilled at a pharmacy without a new office visit in 2008–2010.
3. **Other care settings** — Some patients have **outpatient or inpatient** claims but no carrier claim; drugs may still appear in PDE.
4. **PDE-only beneficiaries** — ~4,700 Sample 1 beneficiaries have Part D fills but **no** carrier, inpatient, or outpatient claim in this dataset (still valid in CMS synthetic data; not everyone with drugs has a matching physician claim in the extract).
5. **Carrier 1A + 1B is not "all medical care"** — it is only ~85% of beneficiaries with any PDE activity, by design of who appears in the carrier split files.

**For joins:** use `DESYNPUF_ID` across files — IDs still map correctly. **For analysis:** do not assume PDE ⊆ Carrier; use left joins and expect PDE-only and carrier-only patients.
